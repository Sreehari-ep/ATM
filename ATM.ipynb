```python
user_data = {
    "pin": "5678",          
    "balance": 10000,        
    "transactions": [],     
    "daily_limit": 50000,
    "withdrawn_today": 0   
}

def authenticate():
    """Authenticate user with PIN"""
    entered_pin = input("Enter your PIN: ")
    if entered_pin == user_data["pin"]:
        print(" Authentication successful!\n")
        return True
    else:
        print(" Incorrect PIN.\n")
        return False


def balance_enquiry():
    """Display current balance"""
    print(f" Your current balance is: {user_data['balance']} INR\n")


def deposit():
    """Deposit money into account"""
    amount = int(input("Enter amount to deposit: "))
    if amount > 0:
        user_data["balance"] += amount
        user_data["transactions"].append(f"Deposited {amount} INR")
        print(f" {amount} INR deposited successfully!\n")
    else:
        print(" Invalid deposit amount.\n")


def withdraw():
    """Withdraw money from account with daily limit check"""
    amount = int(input("Enter amount to withdraw: "))
    if amount <= 0:
        print(" Invalid withdrawal amount.\n")
    elif amount > user_data["balance"]:
        print(" Insufficient balance.\n")
    elif user_data["withdrawn_today"] + amount > user_data["daily_limit"]:
        print(" Daily withdrawal limit exceeded.\n")
    else:
        user_data["balance"] -= amount
        user_data["withdrawn_today"] += amount
        user_data["transactions"].append(f"Withdrew {amount} INR")
        print(f" {amount} INR withdrawn successfully!\n")


def transaction_history():
    """Show all past transactions"""
    if user_data["transactions"]:
        print(" Transaction History:")
        for t in user_data["transactions"]:
            print("-", t)
        print()
    else:
        print("No transactions yet.\n")


def change_pin():
    """Change user PIN"""
    old_pin = input("Enter current PIN: ")
    if old_pin == user_data["pin"]:
        new_pin = input("Enter new PIN: ")
        user_data["pin"] = new_pin
        print(" PIN changed successfully!\n")
    else:
        print(" Incorrect current PIN.\n")

def atm_menu():
    """Main ATM menu loop"""
    if not authenticate():
        return 

    while True:
        print("===== ATM MENU =====")
        print("1. Balance Enquiry")
        print("2. Cash Deposit")
        print("3. Cash Withdrawal")
        print("4. Transaction History")
        print("5. Change PIN")
        print("6. Exit")
        choice = input("Enter your choice (1-6): ")

        if choice == "1":
            balance_enquiry()
        elif choice == "2":
            deposit()
        elif choice == "3":
            withdraw()
        elif choice == "4":
            transaction_history()
        elif choice == "5":
            change_pin()
        elif choice == "6":
            print(" Thank you for using our ATM ")
            break
        else:
            print("Invalid choice. Please try again.\n")
atm_menu()
```
